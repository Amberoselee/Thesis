```python
import pandas as pd
import numpy as np
import csv
import re

from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import Ridge

from transformers import pipeline

# 1. Load Dataset


file_path = "./drive/MyDrive/data/training_set_rel3.tsv"

df = pd.read_csv(
    file_path,
    sep="\t",
    encoding="latin-1",
    engine="python",
    quoting=csv.QUOTE_NONE,
    on_bad_lines="skip"
)
```

```python
# 2. Select essay_set = 1

df = df[df["essay_set"] == 1].copy()
df = df[["essay", "rater1_domain1"]].dropna()

df.rename(columns={
    "essay": "text",
    "rater1_domain1": "score"
}, inplace=True)

df["score"] = df["score"].astype(float)

print("Total samples:", len(df))

df["length"] = df["text"].apply(len)
df["word_count"] = df["text"].apply(lambda x: len(str(x).split()))
print(df["word_count"].describe())
```

```python



df = df[df["word_count"] < 400].copy()


print("\nAfter filtering (<400 words):")
print("Remaining samples:", len(df))
print(df["word_count"].describe())
```

```python
# 3. Detect score range automatically

MIN_SCORE = df["score"].min()
MAX_SCORE = df["score"].max()

print("Score range:", MIN_SCORE, "~", MAX_SCORE)
```

```python
# 4. Split dataset
# training = 10%
# calibration = 60%
# testing = 30%

train_df, remain_df = train_test_split(
    df,
    test_size=0.90,
    random_state=42
)

cal_df, test_df = train_test_split(
    remain_df,
    test_size=1/3,
    random_state=42
)

print("Train size:", len(train_df))
print("Calibration size:", len(cal_df))
print("Test size:", len(test_df))

print("Train ratio:", round(len(train_df) / len(df), 4))
print("Calibration ratio:", round(len(cal_df) / len(df), 4))
print("Test ratio:", round(len(test_df) / len(df), 4))
```

```python
# 5. Load LLM scoring pipeline
import torch
from scipy.sparse import hstack
from transformers import AutoTokenizer, AutoModelForCausalLM

model_name = "Qwen/Qwen3.5-4B"

tokenizer = AutoTokenizer.from_pretrained(model_name)
llm = AutoModelForCausalLM.from_pretrained(
    model_name,
    torch_dtype=torch.float16,
    device_map="auto"
)

device = "cuda" if torch.cuda.is_available() else "cpu"

print("Using device:", device)
```

```python
# 6. Helper functions
def extract_number(text, min_score, max_score):
    nums = re.findall(r"\d+\.?\d*", text)

    if len(nums) == 0:
        return float((min_score + max_score) / 2)

    val = float(nums[0])
    val = max(min_score, min(max_score, val))
    return val


def llm_score(essay, min_score, max_score, n_samples=3):

    scores = []

    for _ in range(n_samples):

        prompt = f"""
You are an expert essay grader.

Score the essay from {min_score} to {max_score} according to the following rubrics.

Score Point 1: An undeveloped response that may take a position but offers no more than very minimal support.
Score Point 2: An under-developed response that may or may not take a position.
Score Point 3: A minimally-developed response that may take a position, but with inadequate support and details.
Score Point 4: A somewhat-developed response that takes a position and provides adequate support.
Score Point 5: A developed response that takes a clear position and provides reasonably persuasive support.
Score Point 6: A well-developed response that takes a clear and thoughtful position and provides persuasive support.

Return only one number.

Essay:
{essay}
"""

        inputs = tokenizer(
            prompt,
            return_tensors="pt",
            truncation=True,
            max_length=512
        ).to(device)

        tokenizer.pad_token = tokenizer.eos_token

        outputs = llm.generate(
            **inputs,
            max_new_tokens=10,
            do_sample=False,
            pad_token_id=tokenizer.eos_token_id
)

        text = tokenizer.decode(
            outputs[0],
            skip_special_tokens=True
        )

        score = extract_number(text, min_score, max_score)
        scores.append(score)

    return np.mean(scores)
```

```python
# 7. Use LLM to score TRAINING set
from tqdm import tqdm
tqdm.pandas()

print("\nScoring training essays with LLM...")
train_df["llm_score"] = train_df["text"].progress_apply(
    lambda x: llm_score(x, MIN_SCORE, MAX_SCORE)
)
```

```python
# 8. Feature Engineering
# TF-IDF text + LLM score
# ============================================================

vectorizer = TfidfVectorizer(
    max_features=3000,
    stop_words="english"
)

X_train_text = vectorizer.fit_transform(train_df["text"])
X_train_num = train_df["llm_score"].values.reshape(-1, 1)

X_train = hstack([X_train_text, X_train_num])
y_train = train_df["score"].values
```

```python
# 9. Train regression model using teacher scores
# ============================================================

model = Ridge(alpha=1.0)
model.fit(X_train, y_train)

print("\nRegression model trained on training set.")
```

```python
# 10. Score CALIBRATION set with LLM
# ============================================================

print("\nScoring calibration essays with LLM...")
cal_df["llm_score"] = cal_df["text"].progress_apply(
    lambda x: llm_score(x, MIN_SCORE, MAX_SCORE)
)

X_cal_text = vectorizer.transform(cal_df["text"])
X_cal_num = cal_df["llm_score"].values.reshape(-1, 1)
X_cal = hstack([X_cal_text, X_cal_num])

cal_pred = model.predict(X_cal)
```

```python
# 11. Alpha list for conformal prediction
import numpy as np

cal_residuals = np.abs(
    cal_df["score"].values - cal_pred
)
alpha_list = [0.01, 0.05, 0.10, 0.15, 0.20, 0.30]
```

```python
# 12. Test Set

print("\nScoring test essays with LLM...")
test_df["llm_score"] = test_df["text"].progress_apply(
    lambda x: llm_score(x, MIN_SCORE, MAX_SCORE)
)

X_test_text = vectorizer.transform(test_df["text"])
X_test_num = test_df["llm_score"].values.reshape(-1, 1)

X_test = hstack([X_test_text, X_test_num])

test_pred = model.predict(X_test)
```

```python
# 13. Prediction Intervals for different alpha values

results = []

for alpha in alpha_list:
    n = len(cal_residuals)

    qhat = np.quantile(
        cal_residuals,
        np.ceil((n + 1) * (1 - alpha)) / n,
        method="higher"
    )

    lower = test_pred - qhat
    upper = test_pred + qhat

    # continuous interval, clipped to valid range
    lower = np.clip(lower, MIN_SCORE, MAX_SCORE)
    upper = np.clip(upper, MIN_SCORE, MAX_SCORE)

    covered = (
        (test_df["score"].values >= lower) &
        (test_df["score"].values <= upper)
    )

    coverage = covered.mean()
    avg_width = np.mean(upper - lower)
    results.append({
        "alpha": alpha,
        "target_coverage": 1 - alpha,
        "qhat": float(qhat),
        "coverage": float(coverage),
        "avg_width": float(avg_width)
    })
```

```python
# 14. Coverage Check

results_df = pd.DataFrame(results)

print("\nCoverage summary:")
print(results_df)

# 15. Point Prediction Error

from sklearn.metrics import mean_absolute_error
mae = mean_absolute_error(
    test_df["score"],
    test_pred
)

print("\nMAE =", round(mae, 4))
```

```python
# 16. Final Results

print("\n==============================")
print("FINAL TEST RESULTS")
print("==============================")

for _, row in results_df.iterrows():
    print(
        f"alpha={row['alpha']:.2f} | "
        f"target={row['target_coverage']:.2f} | "
        f"coverage={row['coverage']:.4f} | "
        f"qhat={row['qhat']:.4f} | "
        f"avg_width={row['avg_width']:.4f}"
    )

print("MAE =", round(mae, 4))
```

```python
# 17. Example Outputs


selected_alpha = 0.30

n = len(cal_residuals)
selected_qhat = np.quantile(
    cal_residuals,
    np.ceil((n + 1) * (1 - selected_alpha)) / n,
    method="higher"
)

test_df["pred_center"] = test_pred
test_df["lower"] = np.clip(test_pred - selected_qhat, MIN_SCORE, MAX_SCORE)
test_df["upper"] = np.clip(test_pred + selected_qhat, MIN_SCORE, MAX_SCORE)

test_df["covered"] = (
    (test_df["score"] >= test_df["lower"]) &
    (test_df["score"] <= test_df["upper"])
)

print("\nSample Predictions:\n")
print(
    test_df[
        [
            "score",
            "pred_center",
            "lower",
            "upper",
            "covered"
        ]
    ].head(20)
)
```

```python
import matplotlib.pyplot as plt

sample_df = test_df[
    ["score", "pred_center", "lower", "upper", "covered"]
].head(20).copy()

sample_df = sample_df.reset_index()
sample_df.rename(columns={"index": "sample_id"}, inplace=True)

fig, ax = plt.subplots(figsize=(10, 8))

y = range(len(sample_df))
xerr = [
    sample_df["pred_center"] - sample_df["lower"],
    sample_df["upper"] - sample_df["pred_center"]
]


ax.errorbar(
    sample_df["pred_center"], y,
    xerr=xerr,
    fmt='o',
    capsize=3
)


for i, row in sample_df.iterrows():
    marker = "s" if row["covered"] else "x"
    ax.scatter(row["score"], i, marker=marker, s=70)

ax.set_yticks(list(y))
ax.set_yticklabels(sample_df["sample_id"].astype(str))
ax.invert_yaxis()
ax.set_xlabel("Score")
ax.set_ylabel("Sample ID")
ax.set_title("Sample Prediction Intervals vs Teacher Scores")
ax.grid(True, axis="x", alpha=0.3)

plt.tight_layout()
plt.show()
```
